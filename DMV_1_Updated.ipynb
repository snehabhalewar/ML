```python
import pandas as pd
import matplotlib.pyplot as plt
from io import StringIO, BytesIO
import json
```

```python
data = {
    "Date": [
        "2026-01-01", "2026-01-02", "2026-01-03",
        "2026-01-04", "2026-01-05", "2026-01-06",
        "2026-01-07", "2026-01-08"
    ],
    "Product": [
        "Laptop", "Mouse", "Keyboard", "Monitor",
        "Mouse", "Laptop", "Keyboard", "Monitor"
    ],
    "Category": [
        "Electronics", "Accessories", "Accessories",
        "Electronics", "Accessories", "Electronics",
        "Accessories", "Electronics"
    ],
    "Quantity": [2, 5, 3, None, 4, 1, 6, 2],
    "Unit_Price": [50000, 500, 1500, 12000, 500, 50000, 1500, 12000],
    "City": [
        "Pune", "Mumbai", "Pune", "Delhi",
        "PunE", "Mumbai", "Pune", "Delhi"
    ]
}
```

```python
csv_text = pd.DataFrame(data).to_csv(index=False)
csv_df = pd.read_csv(StringIO(csv_text))

print(csv_df)
```

```python
excel_buffer = BytesIO()

pd.DataFrame(data).to_excel(
    excel_buffer, index=False, engine="openpyxl"
)

excel_buffer.seek(0)
excel_df = pd.read_excel(excel_buffer)

print(excel_df)
```

```python
json_text = pd.DataFrame(data).to_json(orient="records")
json_df = pd.DataFrame(json.loads(json_text))

print(json_df)
```

```python
print("CSV INFORMATION")
csv_df.info()

print("\nEXCEL INFORMATION")
excel_df.info()

print("\nJSON INFORMATION")
json_df.info()
```

```python
print("CSV:\n\n", csv_df.isnull().sum())
print("\nExcel:\n\n", excel_df.isnull().sum())
print("\nJSON:\n\n", json_df.isnull().sum())
```

```python
print("Duplicate rows in CSV:", csv_df.duplicated().sum())
print("Duplicate rows in Excel:", excel_df.duplicated().sum())
print("Duplicate rows in JSON:", json_df.duplicated().sum())
```

```python
df = pd.concat(
    [csv_df, excel_df, json_df],
    ignore_index=True
)

print(df)
```

```python
df = df.drop_duplicates()

df["Date"] = pd.to_datetime(df["Date"])
df["Quantity"] = pd.to_numeric(df["Quantity"])
df["Unit_Price"] = pd.to_numeric(df["Unit_Price"])

df["Quantity"] = df["Quantity"].fillna(
    df["Quantity"].median()
)

df["City"] = df["City"].replace({"PunE": "Pune"})

df["Total_Sales"] = df["Quantity"] * df["Unit_Price"]

df = df.sort_values("Date")

print(df)
```

```python
print(df[["Quantity", "Unit_Price", "Total_Sales"]].describe())
```

```python
total_sales = df["Total_Sales"].sum()
average_order = df["Total_Sales"].mean()

print("Total Sales = Rs.", total_sales)
print("Average Order Value = Rs.", round(average_order, 2))
```

```python
category_sales = df.groupby("Category")["Total_Sales"].sum()

print(category_sales)
```

```python
city_sales = df.groupby("City")["Total_Sales"].sum()

print(city_sales)
```

```python
category_distribution = df["Category"].value_counts()

print(category_distribution)
```

```python
category_sales.plot(kind="bar", figsize=(7, 5))

plt.title("Total Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales (Rs.)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
category_distribution.plot(
    kind="pie",
    autopct="%1.1f%%",
    figsize=(6, 6)
)

plt.title("Product Category Distribution")
plt.ylabel("")
plt.tight_layout()
plt.show()
```

```python
df["Total_Sales"].plot(kind="box", figsize=(7, 5))

plt.title("Distribution of Total Sales")
plt.ylabel("Total Sales (Rs.)")
plt.tight_layout()
plt.show()
```
